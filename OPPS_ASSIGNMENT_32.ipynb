#  What is Object-Oriented Programming (OOP)
**Object-Oriented Programming (OOP)** is a programming paradigm that organizes software design around **objects**, which are instances of **classes**. These objects can contain **data** (attributes or properties) and **methods** (functions or behaviors).

Here are the core **principles** of OOP:

---

### 🔹 1. **Encapsulation**
- **Definition**: Bundling data and methods that operate on the data within one unit (class), and restricting access to some of the object’s components.
- **Goal**: Hide the internal state and require all interaction to be performed through an object’s methods.
- **Example**: Using private variables and public getters/setters.

---

### 🔹 2. **Abstraction**
- **Definition**: Hiding complex implementation details and showing only the essential features of the object.
- **Goal**: Simplify complex reality by modeling classes appropriate to the problem.

---

### 🔹 3. **Inheritance**
- **Definition**: A mechanism for one class (child/subclass) to inherit properties and behaviors from another class (parent/superclass).
- **Goal**: Promote code reuse and establish a natural hierarchy.

---

### 🔹 4. **Polymorphism**
- **Definition**: The ability for different classes to be treated as instances of the same class through a common interface.
- **Goal**: Write flexible and easily extendable code.

#  What are class and static methods in Python
 Great question! Let's break down **class methods** and **static methods** in Python — both are used in object-oriented programming, but they have different use cases.

---

### 🏫 **Class Method**:

A **class method** is a method that is bound to the **class** and not the **instance** of the class. It can be called on the class itself or on instances of the class, but it always takes the **class** itself as the first argument, usually named `cls`.

Class methods are defined using the `@classmethod` decorator.

#### Key Points:
- Class methods **can modify class-level attributes**.
- They **cannot modify instance-level attributes** unless they are passed an instance.

---

### 🔧 Syntax for Class Method:

```python
class ClassName:
    @classmethod
    def method_name(cls, arguments):
        # Use cls to access class-level attributes or methods
```

### Example of Class Method:

```python
class Dog:
    species = "Canine"  # Class-level attribute

    def __init__(self, name):
        self.name = name  # Instance-level attribute

    @classmethod
    def species_info(cls):
        print(f"All dogs belong to the {cls.species} species.")

# Calling class method
Dog.species_info()  # Output: All dogs belong to the Canine species.
```

---

### 🏁 **Static Method**:

A **static method** is a method that does not take `self` or `cls` as its first parameter. It behaves like a regular function, but it is placed inside the class to indicate that it belongs to that class. It doesn’t modify or access class or instance attributes.

Static methods are defined using the `@staticmethod` decorator.

#### Key Points:
- Static methods do not have access to instance or class attributes.
- They are **independent of the class** and **don’t require an instance** to be called.

---

### 🔧 Syntax for Static Method:

```python
class ClassName:
    @staticmethod
    def method_name(arguments):
        # Independent of class and instance
```

### Example of Static Method:

```python
class Calculator:
    @staticmethod
    def add(x, y):
        return x + y

    @staticmethod
    def multiply(x, y):
        return x * y

# Calling static method without creating an instance
print(Calculator.add(5, 3))        # Output: 8
print(Calculator.multiply(4, 2))   # Output: 8
```

#  What is method overloading in Python
In many programming languages like Java or C++, **method overloading** allows you to define multiple methods with the same name, but with different parameters. Python, however, doesn’t support **method overloading** in the traditional sense, but we can still achieve similar behavior using a few techniques.

---

### 🚫 **Method Overloading in Python (Not Supported Directly)**

In Python, if you define multiple methods with the **same name** in a class, the last one will override the earlier ones. So, there's no direct way to overload methods based on the number or types of arguments.

### 🧠 Example of method overriding (which **replaces** the previous method):

```python
class Example:
    def show(self):
        print("This is the first show method.")

    def show(self):  # This will override the previous show method
        print("This is the second show method.")

# Creating an object
obj = Example()
obj.show()  # Output: This is the second show method.
```

In the example above, the second `show()` method overrides the first one.

---

### ✅ **How to Achieve Method Overloading in Python:**

While Python doesn't support traditional method overloading, you can simulate it using one of these approaches:


- **Method overloading** (as seen in languages like Java or C++) isn't supported directly in Python.
- However, you can **simulate overloading** by:
  - Using **default arguments**.
  - Using `*args` and `**kwargs` for variable arguments.
  - Adding **conditional logic** to handle different argument types or numbers.
  
These techniques allow you to achieve similar behavior to method overloading.

 #  Why is polymorphism important in OOP
 **Polymorphism** is one of the key pillars of Object-Oriented Programming (OOP), and it plays a crucial role in enabling flexibility and extensibility in your code. Let’s break it down to understand why it’s so important.


#### 1. **Flexibility in Code**:
Polymorphism allows you to **write more flexible code** that can work with **objects of multiple classes** in a uniform way, even if those classes implement methods differently.

For example, a **method** that takes an argument of a parent class type can accept any child class object, making it **extensible** and **easy to maintain**.

```python
class Animal:
    def speak(self):
        print("The animal speaks")

class Dog(Animal):
    def speak(self):
        print("Bark!")

class Cat(Animal):
    def speak(self):
        print("Meow!")

def make_animal_speak(animal: Animal):
    animal.speak()

# Polymorphism in action
make_animal_speak(Dog())  # Output: Bark!
make_animal_speak(Cat())  # Output: Meow!
```

- Here, the `make_animal_speak` function works with any subclass of `Animal`. We don’t need to know the specific type of the object — just that it is an `Animal` or subclass.

#### 2. **Code Reusability**:
By using polymorphism, you can **reuse code** that can work with objects of different types. You don’t need to duplicate logic or create specific functions for each class type.

For example, instead of writing separate functions to handle `Dog` and `Cat`, you write a single function that works for any `Animal` object.

#### 3. **Easier to Extend**:
When new classes are added to your program, polymorphism allows you to **extend the functionality** without modifying existing code. If you create a new subclass of `Animal`, it will automatically be compatible with all code that uses polymorphism.

For example, if you added a new class `Bird` that also overrides `speak()`, the existing `make_animal_speak` function would still work.

```python
class Bird(Animal):
    def speak(self):
        print("Tweet!")

make_animal_speak(Bird())  # Output: Tweet!
```

#### 4. **Supports the Open/Closed Principle**:
Polymorphism is closely tied to the **Open/Closed Principle**, one of the SOLID principles in OOP, which states that:
- **Software should be open for extension** (you can add new functionality),
- **But closed for modification** (existing code should remain unchanged).

Polymorphism allows you to extend your system by adding new subclasses without changing the code that uses polymorphism.

#### 5. **Improves Maintainability**:
With polymorphism, you can modify and extend the behavior of specific objects without affecting other parts of the program. If the behavior of a class changes (e.g., changing how `Dog` speaks), you only need to modify the class itself. Other code that interacts with the `Dog` class via polymorphism remains unaffected.

#  What is the difference between a class variable and an instance variable
The difference between a **class variable** and an **instance variable** lies in **how they are stored** and **how they are accessed**. Both are used to store data in Python classes, but they behave differently. Here's a breakdown:

---

### 1. **Class Variable**
- **Definition**: A **class variable** is shared by all instances of the class. It is declared directly inside the class (outside of any methods) and is **not specific to any particular instance** of the class.
- **Scope**: It belongs to the class itself, not to any individual object.
- **Access**: You can access it via the class name or through any instance of the class. Modifying the class variable via an instance will change it for all instances.
- **Use Case**: Class variables are typically used for values that are common to all instances of the class, like constants, default values, or properties that are shared across all instances.

---

### 2. **Instance Variable**
- **Definition**: An **instance variable** is specific to each instance (object) of the class. It is declared inside methods (usually in the `__init__` constructor) and is **associated with the particular instance** of the class.
- **Scope**: It belongs to the instance of the class, so each object can have its own unique set of instance variables.
- **Access**: You can access instance variables via the instance of the class (using `object.variable`), but they are **not shared** between instances.
- **Use Case**: Instance variables are used for data that is specific to each object. For example, the attributes of a `Person` class, like `name` or `age`, would be instance variables because each `Person` object has different values for them.

---

###  **Example of Class Variable and Instance Variable**

```python
class Dog:
    # Class variable: Shared by all instances
    species = "Canis familiaris"  # All dogs belong to the same species
    
    def __init__(self, name, age):
        # Instance variables: Unique to each dog instance
        self.name = name
        self.age = age

# Creating instances of the Dog class
dog1 = Dog("Buddy", 3)
dog2 = Dog("Lucy", 5)

# Accessing class variables
print(dog1.species)  # Output: Canis familiaris
print(dog2.species)  # Output: Canis familiaris

# Modifying class variable via the class itself
Dog.species = "Canis lupus familiaris"

# Accessing class variable after modification
print(dog1.species)  # Output: Canis lupus familiaris
print(dog2.species)  # Output: Canis lupus familiaris

# Accessing instance variables
print(dog1.name)  # Output: Buddy
print(dog1.age)   # Output: 3

# Instance variables are independent of each other
print(dog2.name)  # Output: Lucy
print(dog2.age)   # Output: 5
```

### **Key Points from the Example**:
- **Class Variable (`species`)**:
  - Shared across all instances of the class (`dog1` and `dog2`). If you modify it via the class name, it will change for all instances.
  - You can access the class variable via the class (`Dog.species`) or the instance (`dog1.species`), but it's the same for all objects.
  
- **Instance Variables (`name`, `age`)**:
  - Unique to each instance (`dog1` and `dog2` have different values for `name` and `age`).
  - You access instance variables using the instance itself (`dog1.name`, `dog1.age`).
  - Modifying an instance variable via one object does not affect others.

---

###  **Summary of Differences**:

| Feature                   | **Class Variable**                                    | **Instance Variable**                                  |
|---------------------------|--------------------------------------------------------|--------------------------------------------------------|
| **Scope**                  | Belongs to the class (shared among all instances)     | Belongs to a specific instance (unique to each object) |
| **Access**                 | Can be accessed by the class or any instance          | Accessed via the instance (e.g., `object.variable`)     |
| **Modification**           | Modifying it affects all instances                    | Modifying it only affects the specific instance         |
| **Use Case**               | Common data shared across all instances (constants, default values) | Specific data unique to each instance (e.g., attributes of a person, dog, etc.) |
| **Defined**                | Defined inside the class but outside any methods      | Defined inside the `__init__` method or other instance methods |

---

### 📍 **When to Use Each**:
- **Use class variables** when you want data to be shared among all instances of a class, such as constants, default settings, or values that don't change across objects.
- **Use instance variables** when each object needs to store unique data, such as attributes or properties that are specific to that object.

  #  Explain the purpose of ‘’__str__’ and ‘__repr__’ ‘ methods in Python
In Python, the `__str__` and `__repr__` methods are used to define how an object is represented as a string. They serve different purposes, and understanding their distinction is important for creating readable and useful representations of objects.

### 1. `__str__`: The "user-friendly" string representation

The `__str__` method is used to define how an object is displayed when it is converted to a string (e.g., when you use `print()` or `str()` on an object). It is meant to produce a **human-readable** or **informal** representation of the object that is easy to understand for the user.

#### Purpose:
- The `__str__` method should return a string that represents the object in a way that is easy for humans to understand.
- It is primarily used for displaying the object in a readable and user-friendly manner.

#### Example:
```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} is {self.age} years old."

# Creating an instance of Person
person = Person("Alice", 30)

# Using print() which implicitly calls __str__()
print(person)  # Output: Alice is 30 years old.
```

In this example, when `print(person)` is called, Python uses the `__str__` method to get a human-readable string for the `Person` object.

### 2. `__repr__`: The "developer-friendly" string representation

The `__repr__` method is used to define how an object is represented when it is evaluated in the Python shell or when using `repr()`. It is meant to produce a **developer-friendly** or **formal** representation of the object that can ideally be used to recreate the object.

#### Purpose:
- The `__repr__` method should return a string that, when passed to `eval()`, would ideally recreate the object (though this is not always possible).
- It is intended for debugging and development purposes, providing a detailed and unambiguous representation of the object.

#### Example:
```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

# Creating an instance of Person
person = Person("Alice", 30)

# Using repr() to get the developer-friendly representation
print(repr(person))  # Output: Person('Alice', 30)
```

In this example, the `__repr__` method returns a string that looks like the code needed to create that `Person` object. This helps developers to easily understand the internal state of an object.

### Key Differences:

- **`__str__`** is for a human-readable string representation, primarily used with `print()` or `str()`.
- **`__repr__`** is for a developer-readable string representation, used in the interactive interpreter and debugging. It should aim to provide a more detailed, unambiguous representation of the object.

### The `__repr__` Fallback for `__str__`:

If a class doesn't define `__str__` but does define `__repr__`, then `__str__` will fall back to using `__repr__` when `str()` or `print()` is called. This ensures that there is always a string representation available.

#### Example:
```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

person = Person("Bob", 25)

# __str__ is not defined, so it falls back to __repr__
print(person)  # Output: Person('Bob', 25)
```

### Summary:
- **`__str__`**: Provides a human-readable string representation (for `print()` and `str()`).
- **`__repr__`**: Provides a developer-friendly string representation (for debugging and `repr()`).

# What is the significance of the ‘super()’ function in Python

The `super()` function in Python is used to call methods from a parent class in a subclass. It allows you to refer to the parent class and invoke its methods or properties without explicitly naming the parent class. This is particularly useful in the context of **inheritance**, where a subclass may want to call or override methods from its parent class.

### Significance of `super()`:
1. **Calling Parent Class Methods:**
   `super()` is commonly used to call a method from the superclass (parent class) without directly referring to the superclass by name. This can help avoid tight coupling between the subclass and the parent class, allowing for more maintainable code, especially when dealing with multiple inheritance.

2. **Method Resolution Order (MRO):**
   When you use `super()`, Python automatically determines the correct class to call according to the **Method Resolution Order** (MRO). This is particularly useful in the context of **multiple inheritance**, where a class may inherit from multiple parent classes, and you want Python to automatically figure out which parent class's method should be called.

3. **Avoiding Code Duplication:**
   Using `super()` allows you to avoid code duplication by reusing methods from parent classes. It helps when you want to extend or modify functionality without having to rewrite the entire method in the subclass.

4. **Cooperative Multiple Inheritance:**
   In multiple inheritance scenarios, `super()` supports **cooperative multiple inheritance**, where multiple classes can work together and call each other’s methods in a consistent order based on the MRO.

### Syntax:
```python
super().method_name(arguments)
```

You can also specify which class and instance to refer to using `super(ClassName, instance)`:
```python
super(ClassName, self).method_name(arguments)
```

### Example 1: Simple Inheritance

```python
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):
        super().speak()  # Calls the speak method of the parent class (Animal)
        print("Dog barks")

# Create an instance of Dog
dog = Dog()
dog.speak()
```

**Output:**
```
Animal makes a sound
Dog barks
```

In this example, the `Dog` class overrides the `speak` method, but it still calls the `speak` method from the `Animal` class using `super()`.

### Example 2: Multiple Inheritance

```python
class A:
    def greet(self):
        print("Hello from class A")

class B:
    def greet(self):
        print("Hello from class B")

class C(A, B):
    def greet(self):
        super().greet()  # Calls the greet method from class A (based on MRO)
        print("Hello from class C")

# Create an instance of C
c = C()
c.greet()
```

**Output:**
```
Hello from class A
Hello from class C
```

Here, `super()` automatically follows the **MRO** (Method Resolution Order) and first calls the `greet` method from class `A` before calling the method in class `C`.

### Key Benefits of `super()`:

1. **Cleaner Code:**
   By using `super()`, you avoid having to hard-code the parent class name, making the code easier to maintain and extend, especially when the class hierarchy changes.

2. **Supports Multiple Inheritance:**
   In multiple inheritance, `super()` helps maintain the order of method calls based on the MRO, ensuring that methods from all relevant parent classes are called in a logical order.

3. **Cooperative Inheritance:**
   If multiple parent classes use `super()` themselves, you get **cooperative inheritance**, where each parent class in the hierarchy gets a chance to contribute to the method call, avoiding skipping over methods in the inheritance chain.

### Example: Using `super()` in a Complex Multiple Inheritance Scenario


#  What is the significance of the __del__ method in Python
 In Python, the `__del__` method is a **destructor** method, which is automatically called when an object is about to be destroyed or when it is no longer referenced. Its primary purpose is to perform any necessary cleanup operations before the object is removed from memory.

### Significance of the `__del__` Method:

1. **Resource Cleanup:**
   The `__del__` method is used to release any external resources that the object may be holding onto, such as closing files, network connections, or database connections. It's similar to the concept of "finalizers" in other languages like Java or C++ destructors.

2. **Automatic Cleanup:**
   The `__del__` method is automatically invoked when an object’s reference count reaches zero, meaning there are no more references to the object. This can happen when the object goes out of scope or when it is explicitly deleted using the `del` statement.

3. **Avoiding Resource Leaks:**
   By using `__del__`, you can ensure that resources, such as open files or network connections, are properly closed when the object is no longer needed, preventing resource leaks.

### Syntax:
```python
class MyClass:
    def __del__(self):
        # Cleanup code here
        print("Destructor called, cleaning up...")
```

The `__del__` method is called automatically when the object is about to be destroyed.

### Example of `__del__` Method:

```python
class FileHandler:
    def __init__(self, filename):
        self.filename = filename
        self.file = open(filename, 'w')
        print(f"Opened file: {self.filename}")

    def write(self, data):
        self.file.write(data)

    def __del__(self):
        self.file.close()  # Close the file when the object is destroyed
        print(f"Closed file: {self.filename}")

# Creating an instance of FileHandler

```


In this example, the `__del__` method ensures that the file is closed when the `FileHandler` object is deleted, preventing potential file handle leaks.

### Key Points:

1. **Automatic Invocation:**
   The `__del__` method is invoked automatically when an object is about to be destroyed. You **cannot** call it explicitly; it’s invoked by the garbage collector when it determines that the object is no longer in use.

2. **Unpredictability of `__del__` Calls:**
   - In **Python’s garbage collection model**, the timing of when `__del__` is called is not guaranteed. It depends on when the garbage collector identifies that the object is no longer referenced.
   - This can lead to situations where the `__del__` method is not immediately called or may even be skipped if circular references are involved (e.g., when two objects reference each other).

3. **Circular References:**
   If an object is part of a circular reference (e.g., two objects reference each other), Python’s garbage collector may not be able to immediately destroy the objects, and the `__del__` method may not be called. However, Python has a garbage collector that can detect and clean up circular references, but it might still delay calling `__del__`.

4. **Manual Cleanup:**
   The `__del__` method is typically used for cleanup tasks (e.g., releasing external resources), but it is **not** the best place for critical cleanup logic in all situations. Sometimes, explicitly using a context manager (via the `with` statement) and defining a `__exit__` method is more reliable for resource management (e.g., file handling).

### Example: Handling Circular References and Delaying `__del__`:

```python
        print("A object created")
        
    def __del__(self):

```

```
A object destroyed
```

In this case, the destruction of `A` may be delayed if the object `B` has a circular reference or if Python’s garbage collector does not immediately resolve the object’s references.

### Best Practices:

- **Use Context Managers (`with` statement):** Instead of relying on `__del__` for critical resource management (like file or network connections), use context managers and the `__enter__` and `__exit__` methods to ensure resources are reliably cleaned up.
  
  ```python

  # Using the context manager

      handler.file.write("Hello, World!")
  ```

- **Avoid heavy logic in `__del__`:** The `__del__` method should generally be kept simple and lightweight, as its execution timing is not guaranteed, and it can sometimes introduce unpredictability.

### Conclusion:
The `__del__` method in Python is used for cleaning up resources before an object is destroyed, such as closing files or releasing network connections. However, it has limitations regarding its timing and behavior, especially in the case of circular references, and is generally not the best place to manage critical resources. In practice, context managers are often preferred for resource management.
#  What is the difference between @staticmethod and @classmethod in Python
In Python, both `@staticmethod` and `@classmethod` are used to define methods that are bound to the class rather than to an instance of the class. However, there are key differences between them in terms of how they behave and when to use them.

### Key Differences:

1. **`@staticmethod`:**
   - A **static method** does not take any reference to the class or instance (`self` or `cls`) as its first argument.
   - It behaves like a regular function but resides within a class.
   - Static methods are independent of class and instance state, meaning they don’t have access to or modify class-level attributes or instance-level attributes.
   - They are mainly used for utility functions that are related to the class but do not need access to any instance or class-specific data.

   **Usage:**
   - Use `@staticmethod` when you have a function that logically belongs to a class but does not need to access or modify the class or instance state.

   **Syntax:**
   ```python
   class MyClass:
       @staticmethod
       def static_method(arg1, arg2):
           print("This is a static method.")
   ```

   **Example:**
   ```python
   class MathOperations:
       @staticmethod
       def add(x, y):
           return x + y

   print(MathOperations.add(5, 3))  # Output: 8
   ```

   In this example, `add` is a static method because it doesn't use any instance (`self`) or class (`cls`) attributes. It only operates on the arguments provided.

2. **`@classmethod`:**
   - A **class method** takes a reference to the class (`cls`) as its first argument, which allows it to access and modify class-level attributes, but not instance-level attributes (those bound to `self`).
   - Class methods are bound to the class, not to any specific instance, so they can be called on the class itself or on an instance.
   - Class methods are often used for **alternative constructors** or operations that affect the class as a whole, rather than individual instances.

   **Usage:**
   - Use `@classmethod` when you need access to the class itself (typically to modify class-level data or create alternative constructors).

   **Syntax:**
   ```python
   class MyClass:
       @classmethod
       def class_method(cls, arg1, arg2):
           print("This is a class method.")
   ```

   **Example:**
   ```python
   class Car:
       wheels = 4  # Class attribute

       @classmethod
       def number_of_wheels(cls):
           print(f"A car has {cls.wheels} wheels.")

   # Calling the class method
   Car.number_of_wheels()  # Output: A car has 4 wheels
   ```

   In this example, `number_of_wheels` is a class method because it has access to the class-level attribute `wheels` and can use or modify it.

### Comparison:

| Feature                | `@staticmethod`                          | `@classmethod`                          |
|------------------------|------------------------------------------|-----------------------------------------|
| First argument         | None (`self` or `cls` is not passed)     | `cls` (class reference)                |
| Access to instance    | Cannot access instance (`self`)         | Cannot access instance, but can access class-level attributes (`cls`) |
| Access to class        | Cannot access class (`cls`)             | Can access and modify class-level attributes (`cls`) |
| Purpose                | Utility function, related to class but independent of instance or class state | Often used for class-level operations, such as alternative constructors |
| Example use case       | Helper methods that don't need class or instance data | Factory methods, modifying class attributes, etc. |

### Example of `@staticmethod` and `@classmethod` Together:

```python
class Person:
    population = 0  # Class attribute to track the number of people

    def __init__(self, name):
        self.name = name
        Person.population += 1  # Each time a new instance is created, population increases

    @staticmethod
    def is_adult(age):
        return age >= 18

    @classmethod
    def total_population(cls):
        print(f"Total population: {cls.population}")

# Static method usage:
print(Person.is_adult(20))  # Output: True

# Class method usage:
Person.total_population()  # Output: Total population: 0 (or 1 if you create an instance)
```

In this example:
- `is_adult` is a static method because it does not rely on class or instance data.
- `total_population` is a class method because it accesses the `population` class attribute to print the total number of `Person` objects created.

### When to Use:

- **`@staticmethod`:**
  - When the method does not need access to the class or instance.
  - When the method performs an operation that is logically related to the class but does not need to modify its state.
  
- **`@classmethod`:**
  - When the method needs to access or modify class-level attributes or methods.
  - When you need to define an **alternative constructor** (e.g., creating an instance of a class from a different kind of input).

### Conclusion:

- Use `@staticmethod` for utility functions that don't depend on the instance or the class itself.
- Use `@classmethod` for methods that need to interact with the class itself (not just instances of the class), especially for modifying class-level data or providing alternative constructors.
#  How does polymorphism work in Python with inheritance
### Polymorphism in Python with Inheritance

**Polymorphism** is a core concept in object-oriented programming (OOP) that allows different classes to be treated as instances of the same class through inheritance. In Python, polymorphism means that different classes can define methods with the same name, but these methods can behave differently based on the object (or class) that invokes them.

In the context of inheritance, **polymorphism** allows a subclass to provide a specific implementation of a method that is already defined in its superclass. When a method is called on an object, Python will automatically choose the method that corresponds to the object's actual class, even if the method is defined in the parent class. This is known as **method overriding**.

### Key Points:
- **Overriding:** A subclass can override methods defined in the parent class to provide its own behavior.
- **Dynamic Dispatch:** Python dynamically chooses the appropriate method based on the actual type of the object (not the type of the reference variable).
- **Same Interface, Different Implementation:** Polymorphism enables different classes to share the same method name, but each class can implement the method differently.

### Example of Polymorphism in Python with Inheritance:

```python
class Animal:
    def speak(self):
        print("Animal makes a sound")

class Dog(Animal):
    def speak(self):
        print("Dog barks")

class Cat(Animal):
    def speak(self):
        print("Cat meows")

# Demonstrating Polymorphism
def animal_sound(animal):
    animal.speak()  # The method 'speak' behaves differently based on the object type

# Create objects
dog = Dog()
cat = Cat()

# Passing objects to the same function
animal_sound(dog)  # Output: Dog barks
animal_sound(cat)  # Output: Cat meows
```

### Explanation of the Example:
1. **Superclass `Animal`:** The `Animal` class defines a method `speak()` that is inherited by both `Dog` and `Cat` classes.
2. **Subclasses `Dog` and `Cat`:** Both classes override the `speak()` method to provide their own behavior (dog barks, cat meows).
3. **Polymorphism in Action:** The `animal_sound()` function demonstrates polymorphism. Even though the function is written to handle any object of type `Animal`, it calls the `speak()` method on objects of type `Dog` and `Cat`. The correct method is called based on the object's actual class (`Dog` or `Cat`), not the reference type (`Animal`).

### **Method Resolution Order (MRO) in Python:**
Polymorphism relies heavily on Python's **Method Resolution Order (MRO)**, which defines the order in which methods are looked up in the class hierarchy. Python uses the **C3 linearization algorithm** to determine the order of method lookup.

In multiple inheritance scenarios, Python looks for the method in the base classes in the order defined by the MRO. If the method is found in a base class, it is called, and if not, Python continues searching up the chain of inheritance.

### Example with Multiple Inheritance:

```python
class A:
    def do_something(self):
        print("Class A doing something")

class B(A):
    def do_something(self):
        print("Class B doing something")

class C(A):
    def do_something(self):
        print("Class C doing something")

class D(B, C):
    pass

# Creating an object of class D
d = D()
d.do_something()  # Output: Class B doing something
```

**Explanation:**
- In this example, both classes `B` and `C` override the method `do_something()` from class `A`.
- Class `D` inherits from both `B` and `C`, and when we call `do_something()` on an object of class `D`, the method from `B` is called because Python follows the MRO (Method Resolution Order) and `B` appears before `C` in the order of inheritance (`D(B, C)`).

The MRO can be checked using the `mro()` method:
```python
print(D.mro())  # Output: [<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>]
```

### Benefits of Polymorphism:
1. **Flexibility and Extensibility:** You can add new classes that have a consistent interface with the existing classes, without modifying the existing code that relies on the interface. This makes your code more flexible and easier to extend.
2. **Code Reusability:** Polymorphism allows code to be written in a generic way, so it can work with different types of objects that share the same method names.
3. **Cleaner Code:** Polymorphism helps in writing cleaner and more abstract code by focusing on what actions should be performed, rather than how they should be implemented.

### Summary:
- **Polymorphism** in Python enables the use of methods in different classes that share the same name but behave differently.
- It is mainly achieved through **method overriding** in subclasses, where the subclass provides its own implementation of a method that is defined in the superclass.
- **Polymorphism** is powerful because it allows for writing generic code that can handle objects of different types in a uniform way.

#  What is the purpose of the __call__ method in Python
The `__call__` method in Python is a special (or "magic") method that allows an instance of a class to be called like a function. In other words, when you define a `__call__` method within a class, you can invoke instances of that class as if they were functions. This gives you the ability to create callable objects.

### Purpose of `__call__`:

1. **Making Objects Callable:** The primary purpose of the `__call__` method is to allow instances of a class to be used as if they were functions. By defining this method, you can customize the behavior of an object when it is called directly.
  
2. **Functional Behavior:** This method can be used to implement behavior that is more functional in nature, such as creating objects that can act like functions or as decorators.

3. **Encapsulation of Callable Logic:** Sometimes, you might want to encapsulate certain logic or data inside a class but still want an object to behave like a function for ease of use.

### Syntax of `__call__`:
```python
class MyClass:
    def __call__(self, *args, **kwargs):
        # Do something when the object is called
        print(f"Object called with arguments: {args}, {kwargs}")
```

- The `__call__` method can accept any number of positional (`*args`) and keyword arguments (`**kwargs`), just like a regular function.

### Example of Using `__call__`:

```python
class Adder:
    def __init__(self, increment):
        self.increment = increment

    def __call__(self, value):
        return value + self.increment

# Creating an instance of Adder
add_five = Adder(5)

# Now you can call the object like a function
result = add_five(10)  # This is equivalent to calling add_five.__call__(10)
print(result)  # Output: 15
```

### Breakdown of the Example:
- The `Adder` class defines the `__call__` method, which takes one argument (`value`) and adds the `increment` to it.
- When we create an instance of `Adder` (with `increment=5`), we can call `add_five(10)`, and Python internally invokes `add_five.__call__(10)`, resulting in `10 + 5 = 15`.

### Use Cases for `__call__`:

1. **Creating Callable Objects:**
   When you want your class instance to act like a function, you can define the `__call__` method. For example, in the case of decorators or function factories, the object itself can be called to execute some logic.

2. **Function Factories:**
   You can use the `__call__` method to generate functions dynamically. This is useful when you need a customizable function that is created at runtime based on certain parameters.

   ```python
   class Multiply:
       def __init__(self, factor):
           self.factor = factor

       def __call__(self, value):
           return value * self.factor

   # Create a "function" that multiplies by 3
   multiplier = Multiply(3)
   print(multiplier(4))  # Output: 12
   ```

3. **Creating Decorators:**
   You can use the `__call__` method to define callable objects that work as decorators.

   ```python
   class Timer:
       def __call__(self, func):
           import time
           def wrapper(*args, **kwargs):
               start = time.time()
               result = func(*args, **kwargs)
               print(f"Execution time: {time.time() - start} seconds")
               return result
           return wrapper

   timer = Timer()

   @timer
   def slow_function():
       import time
       time.sleep(2)

   slow_function()  # Will print the execution time
   ```

   In this example, the `Timer` class is used as a decorator. When you decorate a function with `@timer`, Python calls the `__call__` method of the `Timer` object, passing the `slow_function` to it. The `__call__` method returns a new function (`wrapper`) that adds functionality (measuring execution time) around the original function.

4. **Callable Objects for Configuration or Data Processing:**
   Another common use case is to create objects that can be configured with certain parameters and then act like functions to process data. This can be useful when creating configuration-driven behavior.

### Example of Callable Object as a Function Factory:

```python
class PowerOf:
    def __init__(self, exponent):
        self.exponent = exponent

    def __call__(self, base):
        return base ** self.exponent

# Create a callable object that computes squares

print(square(4))  # Output: 16 (4 squared)

# Create a callable object that computes cubes

```

### Summary:

- **`__call__`** allows instances of a class to be called like functions, enabling objects to behave in a functional way.
- By defining the `__call__` method, you can encapsulate logic inside an object while still allowing it to be invoked as if it were a function.
- This is useful for creating **function factories**, **decorators**, and **callable objects** that simplify certain programming tasks, especially when dealing with dynamic behavior or customization.

In short, the `__call__` method in Python allows objects to become more flexible by making them callable, providing a function-like interface while maintaining the full power of class-based encapsulation.

# . Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes
balance and methods to deposit, withdraw, and check balance

````python
class BankAccount:
    def __init__(self, initial_balance=0):
        # Private attribute
        self.__balance = initial_balance

    # Method to deposit money
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited {amount}. New balance is {self.__balance}.")
        else:
            print("Deposit amount must be positive.")

    # Method to withdraw money
    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrew {amount}. New balance is {self.__balance}.")
        elif amount > self.__balance:
            print("Insufficient funds!")
        else:
            print("Withdrawal amount must be positive.")

    # Method to check the balance (getter)
    def check_balance(self):
        print(f"Current balance is {self.__balance}.")

# Create a BankAccount instance
account = BankAccount(1000)  # Initial balance of 1000

# Demonstrating encapsulation by using public methods to interact with private attribute
account.check_balance()  # Output: Current balance is 1000.
account.deposit(500)     # Output: Deposited 500. New balance is 1500.
account.withdraw(200)    # Output: Withdrew 200. New balance is 1300.
account.withdraw(1500)   # Output: Insufficient funds!
account.check_balance()  # Output: Current balance is 1300.
````

#  Create a class MathOperations with a class method add_numbers() to add two numbers and a static
method subtract_numbers() to subtract two numbers.

````python
class MathOperations:
    # Class method to add two numbers
    @classmethod
    def add_numbers(cls, num1, num2):
        return num1 + num2

    # Static method to subtract two numbers
    @staticmethod
    def subtract_numbers(num1, num2):
        return num1 - num2

# Using the class method and static method

# Calling the class method using the class itself
sum_result = MathOperations.add_numbers(5, 3)
print(f"Sum: {sum_result}")  # Output: Sum: 8

# Calling the static method using the class itself
difference_result = MathOperations.subtract_numbers(10, 4)
print(f"Difference: {difference_result}")  # Output: Difference: 6
````

# . Implement a class Person with a class method to count the total number of persons created.

````python
class Person:
    # Class attribute to count the number of Person objects
    person_count = 0

    def __init__(self, name, age):
        self.name = name
        self.age = age
        # Increment the person count each time a new Person is created
        Person.person_count += 1

    # Class method to get the total number of persons created
    @classmethod
    def get_person_count(cls):
        return cls.person_count

# Creating instances of Person
person1 = Person("Alice", 30)
person2 = Person("Bob", 25)
person3 = Person("Charlie", 35)

# Accessing the total count of Person objects created
print(f"Total number of persons created: {Person.get_person_count()}")  # Output: Total number of persons created: 3
````

# . Write a class Fraction with attributes numerator and denominator. Override the str method to display the
fraction as "numerator/denominator".
 class Fraction:
  

````python
class Fraction:
    def __init__(self, numerator, denominator):
        self.numerator = numerator
        self.denominator = denominator

    # Override __str__ to display the fraction in "numerator/denominator" format
    def __str__(self):
        return f"{self.numerator}/{self.denominator}"

# Creating instances of Fraction
fraction1 = Fraction(3, 4)
fraction2 = Fraction(5, 7)

# Printing the fractions using the overridden __str__ method
print(fraction1)  # Output: 3/4
print(fraction2)  # Output: 5/7
````

# . Demonstrate operator overloading by creating a class Vector and overriding the add method to add two
vectors

````python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading the + operator using __add__
    def __add__(self, other):
        if isinstance(other, Vector):  # Ensure the other object is also a Vector
            return Vector(self.x + other.x, self.y + other.y)
        return NotImplemented

    # Method to represent the Vector object as a string
    def __str__(self):
        return f"Vector({self.x}, {self.y})"

# Creating instances of Vector
v1 = Vector(2, 3)
v2 = Vector(4, 5)

# Adding two vectors using the overloaded + operator
v3 = v1 + v2  # Calls the __add__ method

# Printing the result of the addition
print(v1)  # Output: Vector(2, 3)
print(v2)  # Output: Vector(4, 5)
print(v3)  # Output: Vector(6, 8)
````

# 2. Implement a class Student with attributes name and grades. Create a method average_grade() to compute
the average of the grades

````python
class Student:
    def __init__(self, name, grades):
        # Initialize the student's name and grades
        self.name = name
        self.grades = grades

    # Method to calculate the average grade
    def average_grade(self):
        if len(self.grades) == 0:
            return 0  # Avoid division by zero if no grades are provided
        return sum(self.grades) / len(self.grades)

# Creating an instance of Student
student1 = Student("John", [85, 90, 78, 92, 88])

# Calling the average_grade method
avg_grade = student1.average_grade()
print(f"{student1.name}'s average grade is {avg_grade:.2f}")  # Output: John's average grade is 86.60
````

# 3. Create a class Rectangle with methods set_dimensions() to set the dimensions and area() to calculate the
area

````python
class Rectangle:
    def __init__(self):
        # Initialize length and width to None or 0
        self.length = 0
        self.width = 0

    # Method to set the dimensions of the rectangle
    def set_dimensions(self, length, width):
        self.length = length
        self.width = width

    # Method to calculate the area of the rectangle
    def area(self):
        return self.length * self.width

# Creating an instance of Rectangle
rectangle1 = Rectangle()

# Setting the dimensions of the rectangle
rectangle1.set_dimensions(5, 3)

# Calculating and printing the area of the rectangle
print(f"The area of the rectangle is: {rectangle1.area()}")  # Output: The area of the rectangle is: 15
````

#  14. Create a class Employee with a method calculate_salary() that computes the salary based on hours worked
and hourly rate. Create a derived class Manager that adds a bonus to the salary.

````python
# Base class Employee
class Employee:
    def __init__(self, hours_worked, hourly_rate):
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    # Method to calculate the salary based on hours worked and hourly rate
    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate


# Derived class Manager, inheriting from Employee
class Manager(Employee):
    def __init__(self, hours_worked, hourly_rate, bonus):
        # Initialize attributes using the parent class constructor
        super().__init__(hours_worked, hourly_rate)
        self.bonus = bonus

    # Override the calculate_salary method to include the bonus
    def calculate_salary(self):
        base_salary = super().calculate_salary()  # Get base salary from the parent class
        return base_salary + self.bonus  # Add bonus to the base salary


# Creating an instance of Employee
employee1 = Employee(40, 20)  # 40 hours worked at $20/hour

# Creating an instance of Manager
manager1 = Manager(40, 20, 500)  # 40 hours worked at $20/hour + $500 bonus

# Calculating and printing the salary for the Employee
print(f"Employee's salary: ${employee1.calculate_salary()}")  # Output: Employee's salary: $800

# Calculating and printing the salary for the Manager
print(f"Manager's salary: ${manager1.calculate_salary()}")  # Output: Manager's salary: $1300
````

# 15. Create a class Product with attributes name, price, and quantity. Implement a method total_price() that
calculates the total price of the product

````python
class Product:
    def __init__(self, name, price, quantity):
        # Initialize the product's name, price, and quantity
        self.name = name
        self.price = price
        self.quantity = quantity

    # Method to calculate the total price of the product
    def total_price(self):
        return self.price * self.quantity

# Creating an instance of Product
product1 = Product("Laptop", 1000, 3)

# Calculating and printing the total price of the product
print(f"Total price for {product1.name}: ${product1.total_price()}")  # Output: Total price for Laptop: $3000
````

# . Create a class Animal with an abstract method sound(). Create two derived classes Cow and Sheep that
implement the sound() method.

````python
from abc import ABC, abstractmethod

# Abstract base class Animal
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass  # Abstract method to be implemented by subclasses

# Derived class Cow
class Cow(Animal):
    def sound(self):
        return "Moo"

# Derived class Sheep
class Sheep(Animal):
    def sound(self):
        return "Baa"

# Creating instances of Cow and Sheep
cow = Cow()
sheep = Sheep()

# Calling the sound method for each animal
print(f"Cow sound: {cow.sound()}")  # Output: Cow sound: Moo
print(f"Sheep sound: {sheep.sound()}")  # Output: Sheep sound: Baa
````

# . Create a class Book with attributes title, author, and year_published. Add a method get_book_info() that
returns a formatted string with the book's details

````python
class Book:
    def __init__(self, title, author, year_published):
        # Initialize the book's attributes
        self.title = title
        self.author = author
        self.year_published = year_published

    # Method to get the book's information in a formatted string
    def get_book_info(self):
        return f"Title: {self.title}\nAuthor: {self.author}\nYear Published: {self.year_published}"

# Creating an instance of Book
book1 = Book("The Great Gatsby", "F. Scott Fitzgerald", 1925)

# Getting and printing the book's details
print(book1.get_book_info())
````

# . Create a class House with attributes address and price. Create a derived class Mansion that adds an
attribute number_of_rooms.

````python
# Base class House
class House:
    def __init__(self, address, price):
        self.address = address
        self.price = price

    # Method to get details of the house
    def get_house_info(self):
        return f"Address: {self.address}\nPrice: ${self.price}"

# Derived class Mansion, inheriting from House
class Mansion(House):
    def __init__(self, address, price, number_of_rooms):
        # Initialize the base class (House) constructor
        super().__init__(address, price)
        self.number_of_rooms = number_of_rooms

    # Method to get details of the mansion
    def get_mansion_info(self):
        return f"{self.get_house_info()}\nNumber of Rooms: {self.number_of_rooms}"

# Creating an instance of House
house1 = House("123 Main St", 250000)

# Creating an instance of Mansion
mansion1 = Mansion("456 Luxury Rd", 5000000, 10)

# Displaying the information for the house and the mansion
print("House Info:")
print(house1.get_house_info())
print("\nMansion Info:")
print(mansion1.get_mansion_info())
````
